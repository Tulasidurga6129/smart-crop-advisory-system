```python
import pandas as pd

df = pd.read_csv("dataset.csv")

print("Dataset shape:", df.shape)
```

```python
X = df[
    [
        "Crop",
        "Crop_Year",
        "Season",
        "State",
        "Area",
        "Annual_Rainfall",
        "Fertilizer",
        "Pesticide",
        "Avg_Temperature",
        "Max_Temperature",
        "Min_Temperature"
    ]
]

y = df["Yield"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["Crop", "Season", "State"]

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)

print("Encoder created successfully!")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

# Create Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Combine preprocessing and model
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", model)
])

print("Random Forest model created successfully!")
```

```python
# Train the model
pipeline.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
pipeline.fit(X_train, y_train)
```

```python
# Make predictions on test data
y_pred = pipeline.predict(X_test)

print("Predictions completed successfully!")
print("First 10 predictions:")
print(y_pred[:10])
```

```python
# Make predictions on the test data
y_pred = pipeline.predict(X_test)

print("Predictions completed successfully!")
print("\nFirst 10 actual yields:")
print(y_test.head(10).values)

print("\nFirst 10 predicted yields:")
print(y_pred[:10])
```

```python
# Make predictions
y_pred = pipeline.predict(X_test)

print("Predictions completed successfully!\n")

print("Actual Yield        Predicted Yield")
print("-----------------------------------")

for actual, predicted in zip(y_test.head(10).values, y_pred[:10]):
    print(f"{actual:.3f}              {predicted:.3f}")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

# Log-transform the target
y_log = np.log1p(y)

# Split the log-transformed target
X_train_log, X_test_log, y_train_log, y_test_log = train_test_split(
    X,
    y_log,
    test_size=0.20,
    random_state=42
)

# Create a new Random Forest
log_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Create pipeline
log_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", log_model)
])

print("Improved model created successfully!")
```

```python
log_pipeline.fit(X_train_log, y_train_log)

print("Improved model trained successfully!")
```

```python
# Predict log-transformed yield
y_pred_log = log_pipeline.predict(X_test_log)

# Convert predictions back to original yield scale
y_pred_improved = np.expm1(y_pred_log)

print("Improved predictions completed!")
print(y_pred_improved[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate improved model metrics
mae_improved = mean_absolute_error(y_test, y_pred_improved)

rmse_improved = np.sqrt(
    mean_squared_error(y_test, y_pred_improved)
)

r2_improved = r2_score(y_test, y_pred_improved)

print("Improved Model Evaluation")
print("-------------------------")
print(f"MAE  : {mae_improved:.4f}")
print(f"RMSE : {rmse_improved:.4f}")
print(f"R²   : {r2_improved:.4f}")
```
